# Sequential Graph

Three nodes run one after another: `first -> second -> third`.
Each node adds a sentence to `final`.

```python
from typing import TypedDict, List  # Import all the data types we need
from langgraph.graph import StateGraph
```

```python
class AgentState(TypedDict):
    name: str
    age: int
    final: str
    skills: List[str]
```

```python
def first_node(state: AgentState) -> AgentState:
    """Greets the user by name."""
    state['final'] = "Hi " + state['name'] + "! "
    return state
```

```python
def second_node(state: AgentState) -> AgentState:
    """Adds the user's age. str() turns the number into text."""
    state['final'] += "You are " + str(state['age']) + " years old. "
    return state
```

```python
def third_node(state: AgentState) -> AgentState:
    """Adds the user's skills. join() turns the list into text."""
    state['final'] += "You have skills in: " + ", ".join(state['skills'])
    return state
```

```python
graph = StateGraph(AgentState)

graph.add_node("first", first_node)
graph.add_node("second", second_node)
graph.add_node("third", third_node)

graph.set_entry_point("first")      # start here
graph.add_edge("first", "second")   # first -> second
graph.add_edge("second", "third")   # second -> third
graph.set_finish_point("third")     # end here

app = graph.compile()
```

```python
result = app.invoke({"name": "Arjun", "age": 24, "skills": ["Python", "LangGraph", "AI"]})
print(result["final"])
```

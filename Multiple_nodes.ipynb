```python
from typing import  TypedDict , List
from langgraph.graph import StateGraph

## Define the State Schema 
class AgentState(TypedDict) :
    value : List[int]
    name: str
    result: str
```

```python



def process_values(state: AgentState) -> AgentState:
    """Adds up the values and builds a message with the name."""
    state["result"] = f"Hi there {state['name']}! Your sum = {sum(state['value'])}"
    return state
```

```python
graph = StateGraph(AgentState)
graph.add_node("processor", process_values)


graph.set_entry_point("processor") # Set the starting node 
graph.set_finish_point("processor") # Set the ending node 
 
app = graph.compile() # Compiling the whole graph
```

```python
answers = app.invoke({"value": [1, 2, 3, 4, 5], "name": "Arjun"})
print(answers["result"])
```
